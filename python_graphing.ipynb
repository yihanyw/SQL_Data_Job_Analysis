```python
import pandas as pd
import matplotlib.pyplot as plt
```

# 1_top_paying_jobs

```python
top_paying = pd.read_csv("data/1_top_paying_jobs.csv")

top_paying = top_paying.sort_values("salary_year_avg")

plt.figure(figsize=(10, 6))
plt.barh(top_paying["job_title"], top_paying["salary_year_avg"])

plt.xlabel("Average Annual Salary ($)")
plt.ylabel("Job Title")
plt.title("Top 10 Highest-Paying Remote Data Analyst Jobs")

plt.tight_layout()
plt.savefig("graphs/1_top_paying_jobs.png", dpi=300)
plt.show()
```

# 3_top_demanded_skills

```python
top_skills = pd.read_csv("data/3_top_demanded_skills.csv")

plt.figure(figsize=(8, 8))
plt.pie(
    top_skills["demand_count"],
    labels=top_skills["skills"],
    autopct="%1.1f%%"
)

plt.title("Top 5 Skills in High-Paying Data Analyst Jobs")
plt.tight_layout()
plt.savefig("graphs/3_top_demanded_skills.png", dpi=300)
plt.show()
```

# 5_optimal_skills

```python
optimal_skills = pd.read_csv("data/5_optimal_skills.csv")

plt.figure(figsize=(10, 7))
plt.scatter(optimal_skills["demand_count"], optimal_skills["avg_salary"])

plt.xlabel("Demand (Number of Job Postings)")
plt.ylabel("Average Salary ($)")
plt.title("Demand vs. Salary for Data Analyst Skills")

plt.tight_layout()
plt.savefig("graphs/5_optimal_skills.png", dpi=300)
plt.show()
```
